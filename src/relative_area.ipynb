```python
import pandas as pd
import numpy as np
import pickle
```

```python
def get_relative_area(Re, Kavit):
    """
    Предсказывает относительную площадь проходного сечения с использованием суррогатной модели.
    
    Параметры:
    Re (float/int): Число Рейнольдса.
    Kavit (float/int): Число кавитации (значение больше 0).
    model: Загруженная суррогатная модель (KRG).
    
    Возвращает:
    float: Одно число — предсказанное значение относительной площади.
    """
    # Загрузка суррогатной модели для относительной площади
    with open('model/relative_area.pkl', 'rb') as f:
        relative_area = pickle.load(f)
    
    # Формирование входного массива
    inlet_area = np.array([[Re, Kavit]])
    
    # Получение предсказания от модели
    pred_raw = relative_area.predict_values(inlet_area)[0][0]
    
    # Округление (по аналогии с pred * 20 / 20)
    pred_rounded = np.round(pred_raw * 20) / 20
    
    # Ограничение сверху 
    if pred_rounded > 0.5:
        pred_rounded = 0.5
        
    return float(pred_rounded)
```

```python
# --- Пример использования ---
Re = 500000 # от 1e5 до 10e5
Kavit = 10 # значение больше 0
```

```python
result = get_relative_area(Re, Kavit)
print(result)
```
